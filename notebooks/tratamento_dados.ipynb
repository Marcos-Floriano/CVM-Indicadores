```python
import pandas as pd


bpa = pd.read_csv("../data/raw/dfp_cia_aberta_BPA_con_2025.csv", sep=';', encoding='latin-1')

bpa.columns = bpa.columns.str.lower()
bpa['vl_conta'] = bpa['vl_conta'] * 1000
bpa_atual = bpa[bpa['ordem_exerc'] == 'ÚLTIMO']
```

```python
contas_bpa = [
    'Ativo Total',
    'Ativo Circulante',
    'Passivo Circulante',
    'Patrimônio Líquido',
    'Caixa e Equivalentes de Caixa'
]

bpa_filtrado = bpa_atual[bpa_atual['ds_conta'].isin(contas_bpa)]
```

```python
bpa_filtrado.to_csv("../data/dado_tratado/bpa_tratado.csv", index=False)
```
