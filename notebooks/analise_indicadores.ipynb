```python
import pandas as pd

bpa = pd.read_csv("../data/dado_tratado/bpa_tratado.csv")
```

```python
def valor(df, conta):
    return df[df['ds_conta'] == conta]['vl_conta'].sum()
```

```python
ativo_total = valor(bpa, 'Ativo Total')
ativo_circulante = valor(bpa, 'Ativo Circulante')
passivo_circulante = valor(bpa, 'Passivo Circulante')
patrimonio_liquido = valor(bpa, 'Patrimônio Líquido')
caixa = valor(bpa, 'Caixa e Equivalentes de Caixa')

indicadores = {
    'Liquidez Corrente': ativo_circulante / passivo_circulante if passivo_circulante else None,
    'Capital de Giro': ativo_circulante - passivo_circulante,
    'Caixa / Ativo Total': caixa / ativo_total if ativo_total else None,
    'Endividamento': 1 - (patrimonio_liquido / ativo_total) if ativo_total else None
}
```

```python
df_indicadores = pd.DataFrame(indicadores.items(), columns=['Indicador', 'Valor'])
df_indicadores

df_indicadores.to_csv("../data/dado_tratado/indicadores_bpa.csv", index=False)
```
